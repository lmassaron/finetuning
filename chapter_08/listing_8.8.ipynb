### Listing 8.8: Running the GSM8K benchmark using the EleutherAI LM-Eval Harness

```python
import lm_eval
from lm_eval.models.huggingface import HFLM

model_path = "microsoft/phi-3-mini-4k-instruct" 
lm = HFLM(pretrained=model_path, device="auto")

results = lm_eval.simple_evaluate(
    model=lm,
    tasks=["gsm8k"],
    num_fewshot=5, # Standard for GSM8K
    limit=100,     # Remove this limit for a full evaluation run
    batch_size=8
)

gsm8k_score = results["results"]["gsm8k"]["exact_match,strict-match"]
print(f"GSM8K Exact Match Score: {gsm8k_score * 100:.2f}%")
```
