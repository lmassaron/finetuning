### Listing 8.2 : Evaluating the sensitivity of BERTScore and BLEURT to adversarial example pairs 

```python
import evaluate

bertscore = evaluate.load("bertscore")
bleurt = evaluate.load("bleurt", config_name="bleurt-large-512")

tricky_preds = [
    "Paris is NOT the capital of France.",
    "London is the capital of France.",
    "France is the capital of Paris.",
    "Paris is the financial capital of France."
]
tricky_refs = ["Paris is the capital of France."] * len(tricky_preds)

bert_results = bertscore.compute(
    predictions=tricky_preds,
    references=tricky_refs,
    model_type="roberta-large",
)

bleurt_results = bleurt.compute(
    predictions=tricky_preds,
    references=tricky_refs
)

print("--- Tricky Results ---")
for i, pred in enumerate(tricky_preds):
    print(f"\nPrediction: '{pred}'")
    print(f"Reference:  '{tricky_refs[i]}'")
    print(f"BERTScore F1: {bert_results['f1'][i]:.4f}")
    print(f"BLEURT Score: {bleurt_results['scores'][i]:.4f}")
```
